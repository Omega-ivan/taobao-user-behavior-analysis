```python
import pandas as pd
```

```python
#读取样本
file_path = r'C:\Users\mountainxhs\Downloads\UserBehavior.csv\UserBehavior_sample.csv'  
columns = ['user_id', 'item_id', 'category_id', 'behavior_type', 'timestamp']

df = pd.read_csv(file_path, header=None, names=columns)
print('原始样本行数:', len(df))
```

```python
# 时间戳转换
df['datetime'] = pd.to_datetime(df['timestamp'], unit='s', utc=True).dt.tz_convert('Asia/Shanghai')
df['date'] = df['datetime'].dt.date
df['hour'] = df['datetime'].dt.hour
```

```python
#  数据质量检查
print('\n--- 数据质量检查 ---')
print('重复行数:', df.duplicated().sum())
print('缺失值:\n', df.isnull().sum())
print('行为类型分布:\n', df['behavior_type'].value_counts())
```

```python
#去重
df = df.drop_duplicates()
```

```python
# ========== 5. 过滤异常时间 ==========
# 数据集时间范围应为 2017-11-25 至 2017-12-03
df = df[(df['datetime'] >= '2017-11-25') & (df['datetime'] < '2017-12-04')]
```

```python
# ========== 6. 校验行为类型 ==========
valid_behaviors = ['pv', 'fav', 'cart', 'buy']
df = df[df['behavior_type'].isin(valid_behaviors)]
```

```python
# ========== 7. 输出清洗报告 ==========
print('\n--- 清洗报告 ---')
print('清洗后行数:', len(df))
print('独立用户数:', df['user_id'].nunique())
print('独立商品数:', df['item_id'].nunique())
print('独立品类数:', df['category_id'].nunique())
print('行为类型分布:\n', df['behavior_type'].value_counts())
```

```python
# ========== 8. 保存清洗后数据 ==========
output_path = r'C:\Users\mountainxhs\Downloads\UserBehavior.csv\cleaned_user_behavior.csv'   # 改成你的实际路径
df.to_csv(output_path, index=False)
print(f'\n清洗后数据已保存到: {output_path}')
```
